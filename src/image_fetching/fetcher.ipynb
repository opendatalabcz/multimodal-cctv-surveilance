```python
import os

from camera_image_decoder import CameraImageFetcher
```

```python
base_path = "/home/tbenes/multimodal-cctv-surveilance/camera_images/budejovicka"

charles_bride_path = os.path.join(base_path, "charles_bridge"),
budejovicka_path = os.path.join(base_path, "budejovicka"),
holesovicky_path = os.path.join(base_path, "holesovicky")
```

```python
# ⚡ ASYNC - 3 monitoringy SOUČASNĚ (místo postupně)
import asyncio

fetcher = CameraImageFetcher(verbose=True)

async def concurrent_monitoring():
    """Spustí 3 monitoringy současně"""
    
    # Vytvoř 3 async tasky - nyní používáme veřejnou metodu
    task1 = fetcher.monitor_async(
        ["101200", "101201"], 
        interval="15m", 
        duration="1d",
        custom_output_dir=charles_bride_path
    )
    
    task2 = fetcher.monitor_async(
        ["500066", "500067"], 
        interval="15m", 
        duration="1d",
        custom_output_dir=holesovicky_path
    )
    
    task3 = fetcher.monitor_async(
        ["500002"], 
        interval="15m", 
        duration="1d",
        custom_output_dir=budejovicka_path
    )
    
    print("🚀 Spouštím 3 monitoringy SOUČASNĚ...")
    print("📹 Karlův most: kamery 101200, 101201")
    print("📹 Holešovičky: kamery 500066 a 500067") 
    print("📹 Budějovická: kamera 500002")
    
    # Spusť všechny současně
    results = await asyncio.gather(task1, task2, task3)
    
    print("✅ Všechny 3 monitoringy dokončeny!")
    return results

# Spuštění
results = await concurrent_monitoring()
```
